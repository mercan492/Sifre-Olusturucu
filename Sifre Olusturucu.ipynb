```python
import secrets  # Daha güvenli rastgelelik için
import string

def sifre_analiz(sifre):
    """Şifrenin gücünü kontrol eder."""
    uzunluk = len(sifre)
    if uzunluk < 8:
        return "🔴 Zayıf (Çok Kısa)"
    elif uzunluk < 12:
        return "🟡 Orta (Daha uzun olabilir)"
    else:
        return "🟢 Güçlü (Harika!)"

def profesyonel_sifre_uret(uzunluk=16, sembol_olsun=True, rakam_olsun=True):
    # Karakter havuzunu hazırla
    karakterler = string.ascii_letters
    if sembol_olsun:
        karakterler += string.punctuation
    if rakam_olsun:
        karakterler += string.digits

    # Güvenli seçim yap (secrets.choice kullanımı)
    sifre = ''.join(secrets.choice(karakterler) for _ in range(uzunluk))
    return sifre

# --- Uygulama Kısmı ---
print("--- 🛡️ Gelişmiş Şifre Oluşturucuya Hoş Geldiniz ---")

try:
    istenen_uzunluk = int(input("Şifre uzunluğu ne olsun? (Örn: 16): "))
    yeni_sifre = profesyonel_sifre_uret(istenen_uzunluk)
    
    print("\n" + "="*30)
    print(f"Üretilen Şifre: {yeni_sifre}")
    print(f"Güvenlik Durumu: {sifre_analiz(yeni_sifre)}")
    print("="*30)
    
except ValueError:
    print("Lütfen geçerli bir sayı giriniz!")
    
```
